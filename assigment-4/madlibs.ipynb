```python
def marvel_madlib():
    hero_name = input("Marvel Hero Name: ")
    sidekick_name = input("Sidekick's Name: ")
    villain_name = input("Villain's Name: ")
    superpower = input("Superpower: ")
    city = input("City: ")
    vehicle = input("Vehicle (e.g., Jet, Bike): ")
    gadget = input("Gadget: ")
    famous_line = input("Famous Catchphrase: ")
    emotion = input("Emotion: ")
    battle_action = input("Action Verb (e.g., punched, blasted): ")
    adj1 = input("Adjective: ")
    adj2 = input("Adjective: ")

    story = f"""
    In the heart of {city}, chaos erupted once again. The notorious {villain_name} had unleashed a {adj1} weapon
    threatening to destroy everything in sight.

    But fear not — {hero_name}, armed with {superpower} and riding their legendary {vehicle}, flew into action.
    With trusty sidekick {sidekick_name} by their side and their {gadget} in hand, the battle was about to begin.

    "{famous_line}!" yelled {hero_name}, charging toward {villain_name} with unmatched bravery.
    The skies lit up as they {battle_action} their enemy in a clash of epic proportions.

    The city held its breath. The fight was intense. {hero_name} fought with {emotion}, knowing that millions depended on them.
    In the end, good triumphed over evil, and {villain_name} was captured and sent to a high-security prison.

    Citizens cheered, calling {hero_name} the most {adj2} Avenger since the age of Iron Man and Captain America.

    Marvel fans everywhere knew: this was not the end — just the beginning of a new saga.
    """

    print(story)

marvel_madlib()
```
