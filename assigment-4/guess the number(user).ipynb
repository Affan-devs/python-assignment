```python
import random

def ai_number_guesser():
    print("🤖 Welcome to the Number Guesser Game!")
    print("Think of a number between 1 and 100. I’ll try to figure it out!")

    lower_bound = 1
    upper_bound = 100
    guess_count = 0

    while lower_bound <= upper_bound:
        guess = random.randint(lower_bound, upper_bound)
        guess_count += 1

        print(f"My guess is: {guess}")
        response = input("Is it (H)igh, (L)ow, or (C)orrect? ").strip().lower()

        if response == 'h':
            upper_bound = guess - 1
        elif response == 'l':
            lower_bound = guess + 1
        elif response == 'c':
            print(f"🎯 Got it! Your number was {guess}. I guessed it in {guess_count} tries.")
            break
        else:
            print("⚠️ Please respond with 'H', 'L', or 'C'.")

    else:
        print("🤔 Hmm... It seems there's some inconsistency in your answers.")

# Run the game
if __name__ == "__main__":
    ai_number_guesser()
```
